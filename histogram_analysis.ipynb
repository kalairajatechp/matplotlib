```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load built-in dataset
df = sns.load_dataset("tips")

sns.histplot(df['total_bill'], kde=False, bins=20, color='skyblue')
plt.title("Histogram - Total Bill")
plt.savefig("histogram_total_bill.png")
plt.show()
```

```python
sns.kdeplot(df['total_bill'], fill=True, color='green')
plt.title("KDE Plot - Total Bill")
plt.savefig("KDE Plot - Total Bill.png")
plt.show()
```

```python
sns.displot(df['total_bill'], kde=True, color='orange')
plt.title("Displot - Total Bill")
plt.savefig("Displot - Total Bill.png") 
plt.show()
```

```python
sns.ecdfplot(df['total_bill'])
plt.title("ECDF Plot - Total Bill")
plt.savefig("ECDF Plot - Total Bill.png")
plt.show()
```

```python
sns.scatterplot(x='total_bill', y='tip', hue='sex', data=df)
plt.title("Scatter Plot - Total Bill vs Tip")
plt.savefig("Scatter Plot - Total Bill vs Tip.png")
plt.show()
```

```python
sns.lineplot(x='size', y='total_bill', data=df)
plt.title("Line Plot - Size vs Total Bill")
plt.savefig("Line Plot - Size vs Total Bill.png")
plt.show()
```

```python
sns.relplot(x='total_bill', y='tip', hue='sex', kind='scatter', data=df)
plt.title("Relplot - Scatter Example")
plt.savefig("Relplot - Scatter Example.png")
plt.show()
```

```python
sns.barplot(x='sex', y='tip', data=df)
plt.title("Bar Plot - Tip by Sex")
plt.savefig("Bar Plot - Tip by Sex.png")
plt.show()
```

```python
sns.countplot(x='day', data=df)
plt.title("Count Plot - Number of Customers per Day")
plt.savefig("Count Plot - Number of Customers per Day.png")
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=df)
plt.title("Box Plot - Total Bill by Day")
plt.savefig("Box Plot - Total Bill by Day.png") 
plt.show()
```

```python
sns.violinplot(x='day', y='total_bill', data=df)
plt.title("Violin Plot - Total Bill by Day")
plt.savefig("Violin Plot - Total Bill by Day.png")
plt.show()
```

```python
sns.stripplot(x='day', y='tip', data=df, jitter=True)
plt.title("Strip Plot - Tip by Day")
plt.savefig("Strip Plot - Tip by Day.png")
plt.show()
```

```python
sns.swarmplot(x='day', y='tip', data=df)
plt.title("Swarm Plot - Tip by Day")
plt.savefig("Swarm Plot - Tip by Day.png")
plt.show()
```

```python
sns.boxenplot(x='day', y='total_bill', data=df)
plt.title("Boxen Plot - Total Bill by Day")
plt.savefig("Boxen Plot - Total Bill by Day.png")
plt.show()
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heatmap - Correlation Matrix")
plt.savefig("Heatmap - Correlation Matrix.png")
plt.show()
```

```python
sns.regplot(x='total_bill', y='tip', data=df)
plt.title("Regression Plot - Total Bill vs Tip")
plt.savefig("Regression Plot - Total Bill vs Tip.png")  
plt.show()
```

```python
sns.lmplot(x='total_bill', y='tip', hue='sex', data=df)
plt.title("LM Plot - Regression by Gender")
plt.savefig("LM Plot - Regression by Gender.png")
plt.show()
```
